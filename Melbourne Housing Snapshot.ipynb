```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv(r"C:\Users\Yadu krishna.M\Downloads\archive (1)\melb_data.csv")
```

```python
print("Original DataSet:")
print(df.head())
```

```python
print("\nShape of the Dataset:")
print(df.shape)
print("\nColumn names")
print(df.columns)
```

```python
print("\ndataset info\n")
print(df.info)

print("\nchecking missing value\n")
print(df.isnull())

print("\nchecking missing value\n")
print(df.isnull().sum())

#creating copy of dataset
df_mean = df.copy()

#creating copy of dataset
df_median=df.copy()
```

```python
#calculate the mean of BuildingArea
mean_building_area=df_mean["BuildingArea"].mean()

#replacing missing value with mean
df_mean["BuildingArea"]=df_mean["BuildingArea"].fillna(mean_building_area)
print("\nMean building Area : ",mean_building_area)
print("missing BuildingArea values after Mean Imputation: ")
print(df_mean["BuildingArea"].isnull().sum())
```

```python
#calculating median of YearBuilt
median_year_built = df_mean["YearBuilt"].median()

#replacing missing value with median
df_median["YearBuilt"]=df_median["YearBuilt"].fillna(median_year_built)
print("\nMedian year built : ",median_year_built)
print("missing yearbuilt values after Median Imputation: ")
print(df_median["YearBuilt"].isnull().sum())
```

```python
df_ffill = df.copy()
df_ffill["Car"] = df_ffill["Car"].ffill()
print("\nMissing car values after forward fill\n")
print(df_ffill["Car"].isnull().sum())
```

```python
comparison = pd.DataFrame({
    "Original BuildingArea": df["BuildingArea"].head(15),
    "Mean Imputed BuildingArea": df_mean["BuildingArea"].head(15),

    "Original YearBuilt": df["YearBuilt"].head(15),
    "Mean Imputed YearBuilt": df_median["YearBuilt"].head(15),

    "Original Car": df["Car"].head(15),
    "Mean Imputed Car": df_ffill["Car"].head(15)
})

print("\nComparision of Imputation Methods:")
print(comparison)
```

```python
#create a copy of the dataset
df_processed =df.copy()
# find the most frequent calue of Councilarea
council_mode = df_processed["CouncilArea"].mode()[0]
#replace missing councilarea values with the mode
df_processed["CouncilArea"] = df_processed["CouncilArea"].fillna(
    council_mode
)

print("\nCouncilArea:",council_mode)

print("\nMissing Catagorical Values after Imputation:")

print(df_processed[["CouncilArea"]].isnull().sum())
```

```python
# convert type and method into numerical colummns
df_encoded = pd.get_dummies(
    df_processed,
    columns=["Type","Method"],
    dtype=int
)

print("\nDataset After One-Hot Encoding:")
print(df_encoded.head())
```

```python
# Lable encoding

df_label = df.copy()

#create label encoder
label_encoder = LabelEncoder()

#Encode the council area column
df_label["CouncilArea_Encoded"] = label_encoder.fit_transform(
    df_label["CouncilArea"]
)

print("\nLabel Encoding of CouncilArea:")
print(
    df_label[
        ["CouncilArea_Encoded"] = label_encoder.fit_transform(
    df_label["CouncilArea"]
)

print("\nLabel Encoding of CouncilArea:")
print(
    df_label[["CouncilArea","CouncilArea_Encoded"]].head(10)
)
```
