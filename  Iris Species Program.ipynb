```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\Yadu krishna.M\Downloads\archive\Iris.csv")
```

```python
print("First 5 rows of the dataset:")
print(df.head())
print("\nLast 5 rows of the dataset:")
print(df.tail())
print("\nShape of the dataset:")
print(df.shape)
print("\nColumns names:")
print(df.columns)
```

```python
print("\nDataset information:")
df.info()
```

```python
print(df.isnull().sum())
df = df.dropna()
print(df.isnull().sum())
```

```python
print(df.duplicated().sum())
df =df.drop_duplicates()
print(df.duplicated().sum())
```

```python
#Creating a Histogram
plt.figure(figsize=(8,5))
sns.histplot(data=df,x="SepalLengthCm",bins=10)
plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Number of Flowers")
plt.show()
```

```python
plt.figure(figsize=(8,5))
#Create a boxplot
sns.boxplot(data=df,x="Species",y="SepalLengthCm")
#ADD a title
plt.title("Sepal Length by Species")
#Add X and Y axis
plt.xlabel("Species")
plt.ylabel("Sepal Length (cm)")
#Display
plt.show()
```

```python
#Create a graph area
plt.figure(figsize=(8,5))
#Create scatter plot
sns.scatterplot(
    data=df,
    x="SepalLengthCm",
    y="PetalLengthCm",
    hue="Species"
)

plt.title("Sepal Length vs Petal Length")
plt.xlabel("Petal Length (Cm)")
plt.ylabel("Sepal Length (Cm)")
plt.show()
```

```python
#create a pair plot
sns.pairplot(df, hue="Species")
#Not needed for the lab, was just experimenting
sns.set_theme(style="darkgrid") 
#display graph
plt.show()
```

```python
#Correlation

#Select the numerical columns
numeric_columns = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]

#calculate correlation
correlation = df[numeric_columns].corr()
#display correlation matrix
print(correlation)
```

```python
#Heat Map
plt.figure(figsize=(8,6))
sns.heatmap(
    correlation,
    annot=True,
    cmap="viridis_r",
    #cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap")
plt.show()
```
